# STRUKTUR R–T–C–I–K–O–Q–V

Struktur R–T–C–I–K–O–Q–V merupakan kerangka yang digunakan untuk
membuat prompt AI agar lebih jelas, terarah, dan sesuai dengan
kebutuhan pengguna. Struktur ini membantu AI memahami peran yang
harus dilakukan, tugas yang diberikan, konteks permasalahan,
instruksi pengerjaan, pengetahuan dan batasan, bentuk keluaran,
pertanyaan utama, serta cara memvalidasi hasil.

| Struktur | Arti | Fungsi |
|---|---|---|
| **R** | **Role** | Menentukan peran atau keahlian AI |
| **T** | **Task** | Menentukan tugas yang harus dilakukan |
| **C** | **Context** | Memberikan latar belakang dan informasi masalah |
| **I** | **Instruction** | Memberikan langkah atau instruksi pengerjaan |
| **K** | **Knowledge/Constraints** | Menentukan pengetahuan dan batasan |
| **O** | **Output** | Menentukan bentuk hasil yang diinginkan |
| **Q** | **Question** | Menentukan pertanyaan utama |
| **V** | **Validation** | Memeriksa kembali hasil yang diperoleh |

# PROMPT 1 — GEODINAMIKA DAN TEKTONOFISIKA

7. Jelaskan hubungan arah pergerakan lempeng dengan jenis batas

9. Jelaskan proses subduksi secara sederhana.

- gempa bumi;

Jangan membuat data pergerakan lempeng yang tidak diberikan.

Jika arah atau kecepatan pergerakan tidak tersedia, jelaskan data
yang diperlukan.

Jangan menyatakan bahwa suatu wilayah pasti mengalami gempa atau

Bagaimana pergerakan lempeng tektonik dan jenis batas lempeng dapat
