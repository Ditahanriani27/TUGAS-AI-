```python

# ============================================================
# ANALISIS PERGERAKAN LEMPENG TEKTONIK
# GEODINAMIKA DAN TEKTONOFISIKA
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. DATA CONTOH
# Semua angka berikut bersifat ILUSTRATIF, bukan data observasi.
# Arah dinyatakan sebagai azimut: 0=N, 90=E, 180=S, 270=W.
# Koordinat merupakan titik contoh untuk visualisasi.
# ------------------------------------------------------------

data = [
    {
        "Batas": "Batas A",
        "Lempeng 1": "Lempeng Samudra A",
        "Lempeng 2": "Lempeng Benua A",
        "Latitude": -10.0,
        "Longitude": 120.0,
        "Arah 1 (derajat)": 90,
        "Kecepatan 1 (cm/tahun)": 5.0,
        "Arah 2 (derajat)": 270,
        "Kecepatan 2 (cm/tahun)": 2.0,
        "Jenis batas": "Konvergen",
        "Lempeng menunjam": "Lempeng Samudra A"
    },
    {
        "Batas": "Batas B",
        "Lempeng 1": "Lempeng C",
        "Lempeng 2": "Lempeng D",
        "Latitude": 0.0,
        "Longitude": 150.0,
        "Arah 1 (derajat)": 90,
        "Kecepatan 1 (cm/tahun)": 3.0,
        "Arah 2 (derajat)": 270,
        "Kecepatan 2 (cm/tahun)": 3.0,
        "Jenis batas": "Divergen",
        "Lempeng menunjam": "Tidak ditentukan"
    },
    {
        "Batas": "Batas C",
        "Lempeng 1": "Lempeng E",
        "Lempeng 2": "Lempeng F",
        "Latitude": 10.0,
        "Longitude": 170.0,
        "Arah 1 (derajat)": 0,
        "Kecepatan 1 (cm/tahun)": 2.0,
        "Arah 2 (derajat)": 0,
        "Kecepatan 2 (cm/tahun)": 2.0,
        "Jenis batas": "Transform",
        "Lempeng menunjam": "Tidak ditentukan"
    }
]

df = pd.DataFrame(data)

# ------------------------------------------------------------
# 2. FUNGSI KOMPONEN VEKTOR
# Azimut diukur searah jarum jam dari utara.
# Komponen timur = v sin(azimut)
# Komponen utara = v cos(azimut)
# ------------------------------------------------------------

def komponen(arah_derajat, kecepatan):
    sudut = np.radians(arah_derajat)
    timur = kecepatan * np.sin(sudut)
    utara = kecepatan * np.cos(sudut)
    return timur, utara

# Hitung komponen gerak kedua lempeng
hasil = []

for _, r in df.iterrows():
    e1, n1 = komponen(
        r["Arah 1 (derajat)"],
        r["Kecepatan 1 (cm/tahun)"]
    )
    e2, n2 = komponen(
        r["Arah 2 (derajat)"],
        r["Kecepatan 2 (cm/tahun)"]
    )

    # Kecepatan relatif lempeng 1 terhadap lempeng 2
    de = e1 - e2
    dn = n1 - n2
    vrel = np.hypot(de, dn)

    # Azimut relatif: 0=N, 90=E, 180=S, 270=W
    azimut_rel = np.degrees(np.arctan2(de, dn)) % 360

    hasil.append({
        "Komponen Timur Relatif": de,
        "Komponen Utara Relatif": dn,
        "Kecepatan Relatif (cm/tahun)": vrel,
        "Azimut Relatif (derajat)": azimut_rel
    })

df_hasil = pd.concat(
    [df, pd.DataFrame(hasil)], axis=1
)

# ------------------------------------------------------------
# 3. VALIDASI DATA
# ------------------------------------------------------------

print("=" * 65)
print("ANALISIS GEODINAMIKA DAN TEKTONOFISIKA")
print("=" * 65)
print("\nCATATAN: seluruh data pada contoh ini bersifat ilustratif.")

kolom_tabel = [
    "Batas", "Lempeng 1", "Lempeng 2",
    "Latitude", "Longitude",
    "Arah 1 (derajat)", "Kecepatan 1 (cm/tahun)",
    "Arah 2 (derajat)", "Kecepatan 2 (cm/tahun)",
    "Jenis batas", "Lempeng menunjam"
]

print("\n3. DATA YANG DIGUNAKAN")
print(df[kolom_tabel].to_string(index=False))

# ------------------------------------------------------------
# 4. PENGERTIAN DASAR
# ------------------------------------------------------------

print("\n1. PENGERTIAN GEODINAMIKA DAN TEKTONOFISIKA")
print(
    "Geodinamika mempelajari proses dan gerakan bumi, termasuk "
    "pergerakan lempeng serta evolusi bagian dalam bumi."
)
print(
    "Tektonofisika mempelajari proses fisika pada deformasi kerak "
    "dan litosfer, termasuk tegangan, regangan, dan patahan."
)

print("\n2. KONSEP DASAR LEMPENG TEKTONIK")
print(
    "Lempeng tektonik adalah bagian kaku litosfer yang bergerak "
    "di atas astenosfer yang dapat mengalami deformasi."
)

print("\n4. JENIS BATAS LEMPENG")
print(
    "- Konvergen: dua lempeng saling mendekat.\n"
    "- Divergen: dua lempeng saling menjauh.\n"
    "- Transform: dua lempeng bergeser mendatar relatif satu sama lain."
)

# ------------------------------------------------------------
# 5. ARAH DAN KECEPATAN
# ------------------------------------------------------------

print("\n5. ARAH DAN KECEPATAN PERGERAKAN")
print(
    df_hasil[[
        "Batas",
        "Arah 1 (derajat)", "Kecepatan 1 (cm/tahun)",
        "Arah 2 (derajat)", "Kecepatan 2 (cm/tahun)",
        "Kecepatan Relatif (cm/tahun)",
        "Azimut Relatif (derajat)"
    ]].round(2).to_string(index=False)
)

# ------------------------------------------------------------
# 6. ANALISIS PERGERAKAN LEMPENG
# ------------------------------------------------------------

print("\n6. ANALISIS PERGERAKAN LEMPENG")

for _, r in df_hasil.iterrows():
    print(
        f'{r["Batas"]}: kecepatan relatif '
        f'{r["Kecepatan Relatif (cm/tahun)"]:.2f} cm/tahun, '
        f'azimut relatif '
        f'{r["Azimut Relatif (derajat)"]:.1f} derajat.'
    )

# ------------------------------------------------------------
# 7. IDENTIFIKASI AWAL ZONA SUBDUKSI
# ------------------------------------------------------------

print("\n7. IDENTIFIKASI ZONA SUBDUKSI")

for _, r in df_hasil.iterrows():
    if r["Jenis batas"] == "Konvergen":
        if r["Lempeng menunjam"] != "Tidak ditentukan":
            print(
                f'{r["Batas"]}: kandidat zona subduksi dalam '
                f'data contoh. Lempeng yang diasumsikan menunjam: '
                f'{r["Lempeng menunjam"]}. Perlu bukti geologi '
                f'untuk memverifikasi.'
            )
        else:
            print(
                f'{r["Batas"]}: konvergen, tetapi lempeng yang '
                f'menunjam belum diketahui.'
            )
    else:
        print(
            f'{r["Batas"]}: jenis batas {r["Jenis batas"]}; '
            f'tidak diidentifikasi sebagai zona subduksi dari '
            f'informasi ini saja.'
        )

# ------------------------------------------------------------
# 8. PROSES GEODINAMIKA
# ------------------------------------------------------------

print("\n8. PROSES GEODINAMIKA")
print(
    "Pada batas konvergen, lempeng dapat bertumbukan. Jika satu "
    "lempeng menunjam ke bawah lempeng lainnya, terbentuk zona "
    "subduksi. Proses ini dapat menyebabkan deformasi, pembentukan "
    "palung, dan perubahan kondisi batuan."
)

# ------------------------------------------------------------
# 9. FENOMENA GEOFISIKA
# ------------------------------------------------------------

print("\n9. FENOMENA GEOFISIKA YANG BERKAITAN")
print(
    "- Gempa bumi: dapat terjadi akibat pelepasan tegangan pada "
    "patahan atau bidang kontak lempeng.\n"
    "- Aktivitas vulkanik: dapat berkaitan dengan pembentukan magma "
    "pada sistem subduksi tertentu.\n"
    "- Deformasi kerak: dapat diamati menggunakan pengukuran GPS/GNSS.\n"
    "Jenis batas lempeng saja tidak dapat memastikan waktu atau "
    "besarnya gempa maupun letusan."
)

# ------------------------------------------------------------
# 10. INTERPRETASI
# ------------------------------------------------------------

print("\n10. INTERPRETASI")
for _, r in df_hasil.iterrows():
    jenis = r["Jenis batas"]
    if jenis == "Konvergen":
        keterangan = (
            "gerak mendekat; subduksi mungkin terjadi jika "
            "ada lempeng yang menunjam"
        )
    elif jenis == "Divergen":
        keterangan = "gerak menjauh; kerak baru dapat terbentuk"
    else:
        keterangan = (
            "pergeseran mendatar; deformasi geser dapat terjadi"
        )

    print(f'{r["Batas"]}: {keterangan}.')

# ------------------------------------------------------------
# 11. KETERBATASAN
# ------------------------------------------------------------

print("\n11. KETERBATASAN")
print(
    "Data ini dibuat untuk demonstrasi kode, bukan pengukuran "
    "pergerakan lempeng yang sebenarnya. Koordinat hanya titik "
    "contoh. Penetapan jenis batas dan lempeng yang menunjam harus "
    "diperiksa dengan peta tektonik dan data ilmiah. Kecepatan "
    "relatif dihitung dengan asumsi kedua vektor berada dalam "
    "kerangka acuan yang sama."
)

# ------------------------------------------------------------
# 12. KESIMPULAN
# ------------------------------------------------------------

print("\n12. KESIMPULAN")
print(
    "Arah, kecepatan, dan jenis batas dapat digunakan untuk "
    "menafsirkan pola gerak lempeng. Batas konvergen merupakan "
    "lokasi yang perlu diperiksa lebih lanjut untuk menentukan "
    "adanya subduksi. Kesimpulan geologi nyata membutuhkan data "
    "yang tervalidasi."
)

# ------------------------------------------------------------
# 13. VISUALISASI VEKTOR
# Panah menunjukkan arah dan besar vektor relatif.
# Grafik ini memakai koordinat ilustratif, bukan peta dasar.
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(10, 6))

warna = {
    "Konvergen": "red",
    "Divergen": "green",
    "Transform": "blue"
}

for _, r in df_hasil.iterrows():
    e = r["Komponen Timur Relatif"]
    n = r["Komponen Utara Relatif"]

    ax.scatter(
        r["Longitude"], r["Latitude"],
        color=warna[r["Jenis batas"]], s=70
    )

    ax.quiver(
        r["Longitude"], r["Latitude"],
        e, n,
        angles="uv",
        scale_units="inches",
        scale=2,
        color=warna[r["Jenis batas"]],
        width=0.006
    )

    ax.text(
        r["Longitude"] + 1.5,
        r["Latitude"] + 1.0,
        r["Batas"],
        fontsize=9
    )

ax.set_title(
    "Vektor Gerak Relatif Lempeng (Data Ilustratif)"
)
ax.set_xlabel("Longitude contoh (derajat)")
ax.set_ylabel("Latitude contoh (derajat)")
ax.grid(True, alpha=0.3)

from matplotlib.lines import Line2D

legenda = [
    Line2D([0], [0], color="red", marker="o",
           label="Konvergen", linestyle="None"),
    Line2D([0], [0], color="green", marker="o",
           label="Divergen", linestyle="None"),
    Line2D([0], [0], color="blue", marker="o",
           label="Transform", linestyle="None")
]

ax.legend(handles=legenda)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 14. GRAFIK PERBANDINGAN KECEPATAN RELATIF
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))
plt.bar(
    df_hasil["Batas"],
    df_hasil["Kecepatan Relatif (cm/tahun)"]
)
plt.ylabel("Kecepatan relatif (cm/tahun)")
plt.xlabel("Batas lempeng contoh")
plt.title("Perbandingan Kecepatan Relatif (Ilustratif)")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 15. SIMPAN TABEL HASIL
# ------------------------------------------------------------

df_hasil.to_csv(
    "hasil_analisis_lempeng.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nFile berhasil disiapkan: hasil_analisis_lempeng.csv")
```

```python

# ============================================================
# ANALISIS SPASIAL POTENSI KERAWANAN LONGSOR MENGGUNAKAN GIS
# Metode: Slope + Reclassification + Weighted Overlay
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. PENGATURAN
# ------------------------------------------------------------

# True = data simulasi untuk latihan
# False = membaca raster GIS milik sendiri
USE_DEMO = True

# Bobot contoh, total harus 100%
weights = {
    "slope": 0.30,
    "rainfall": 0.25,
    "landuse": 0.20,
    "soil": 0.15,
    "elevation": 0.10
}

assert np.isclose(sum(weights.values()), 1.0), \
    "Jumlah bobot harus 100%"

# ------------------------------------------------------------
# 2. DATA INPUT
# DEM menggunakan satuan meter dan ukuran sel dalam meter.
# Data simulasi hanya untuk demonstrasi.
# ------------------------------------------------------------

if USE_DEMO:
    np.random.seed(42)
    rows, cols = 100, 100
    cell_size = 30.0  # meter, contoh resolusi raster

    y, x = np.mgrid[0:rows, 0:cols]

    # DEM simulasi (meter)
    dem = (
        400
        + 100 * np.sin(x / 12)
        + 80 * np.cos(y / 15)
        + 30 * np.sin((x + y) / 8)
    )

    dem += np.random.normal(0, 3, (rows, cols))

    # Curah hujan simulasi (mm/tahun)
    rainfall = (
        1800 + 1000 * (y / rows)
        + np.random.normal(0, 100, (rows, cols))
    )
    rainfall = np.clip(rainfall, 500, None)

    # Kode penggunaan lahan:
    # 1 = hutan, 2 = pertanian, 3 = permukiman,
    # 4 = lahan terbuka
    landuse = np.random.choice(
        [1, 2, 3, 4],
        size=(rows, cols),
        p=[0.40, 0.30, 0.20, 0.10]
    )

    # Kode jenis tanah:
    # 1 = relatif stabil, 2 = sedang, 3 = lebih rentan
    soil = np.random.choice(
        [1, 2, 3],
        size=(rows, cols),
        p=[0.35, 0.40, 0.25]
    )

else:
    # Ganti nama file dengan raster milik sendiri.
    # Semua raster harus disamakan grid, CRS, dan resolusinya.
    import rasterio

    dem_path = "dem.tif"
    rainfall_path = "curah_hujan.tif"
    landuse_path = "penggunaan_lahan.tif"
    soil_path = "jenis_tanah.tif"

    with rasterio.open(dem_path) as src:
        dem = src.read(1).astype(float)
        profile = src.profile.copy()
        transform = src.transform
        crs = src.crs
        cell_x = abs(transform.a)
        cell_y = abs(transform.e)

        if src.nodata is not None:
            dem[dem == src.nodata] = np.nan

    def read_aligned(path, categorical=False):
        from rasterio.warp import reproject, Resampling

        with rasterio.open(path) as src:
            out = np.full(dem.shape, np.nan, dtype=np.float32)

            reproject(
                source=rasterio.band(src, 1),
                destination=out,
                src_transform=src.transform,
                src_crs=src.crs,
                dst_transform=transform,
                dst_crs=crs,
                resampling=(
                    Resampling.nearest if categorical
                    else Resampling.bilinear
                ),
                src_nodata=src.nodata,
                dst_nodata=np.nan
            )
        return out

    rainfall = read_aligned(rainfall_path)
    landuse = read_aligned(landuse_path, categorical=True)
    soil = read_aligned(soil_path, categorical=True)

    # Slope akan dihitung dari DEM menggunakan ukuran sel aktual.
    cell_size = None

# ------------------------------------------------------------
# 3. PEMERIKSAAN DATA
# ------------------------------------------------------------

print("=" * 60)
print("ANALISIS SPASIAL POTENSI KERAWANAN LONGSOR")
print("=" * 60)

print("\n1. TUJUAN ANALISIS")
print(
    "Menghasilkan zonasi potensi kerawanan longsor "
    "berdasarkan kombinasi parameter lingkungan."
)

print("\n2. DATA INPUT")
print("DEM, curah hujan, penggunaan lahan, dan jenis tanah.")
print("Ukuran raster:", dem.shape)

if not USE_DEMO:
    print("CRS:", crs)
    print("Ukuran sel:", cell_x, "x", cell_y, "meter")
    if crs is None or not crs.is_projected:
        print(
            "PERINGATAN: gunakan CRS proyeksi yang sesuai "
            "untuk menghitung slope dalam satuan meter."
        )

# ------------------------------------------------------------
# 4. PEMBUATAN SLOPE DARI DEM
# ------------------------------------------------------------

if USE_DEMO:
    dy, dx = np.gradient(dem, cell_size, cell_size)
else:
    dy, dx = np.gradient(dem, cell_y, cell_x)

slope = np.degrees(np.arctan(np.sqrt(dx**2 + dy**2)))

print("\n3. FUNGSI PARAMETER")
print("- DEM: elevasi permukaan.")
print("- Slope: kemiringan lereng hasil pengolahan DEM.")
print("- Hujan: berkaitan dengan kejenuhan tanah.")
print("- Penggunaan lahan: memengaruhi perlindungan tanah.")
print("- Jenis tanah: memengaruhi kekuatan dan permeabilitas.")

# ------------------------------------------------------------
# 5. RECLASSIFICATION
# Semua parameter diubah menjadi skor 1 sampai 5.
# Skor tinggi menunjukkan karakteristik relatif lebih rentan.
# Ambang ini contoh dan harus disesuaikan dengan wilayah.
# ------------------------------------------------------------

def classify_slope(a):
    return np.select(
        [a < 8, a < 15, a < 25, a < 40],
        [1, 2, 3, 4],
        default=5
    ).astype(float)

def classify_rainfall(a):
    return np.select(
        [a < 1500, a < 2000, a < 2500, a < 3000],
        [1, 2, 3, 4],
        default=5
    ).astype(float)

slope_score = classify_slope(slope)
rain_score = classify_rainfall(rainfall)

# Skor penggunaan lahan: sesuaikan kode kelas dengan data asli.
# 1=hutan, 2=pertanian, 3=permukiman, 4=lahan terbuka
land_score = np.select(
    [landuse == 1, landuse == 2,
     landuse == 3, landuse == 4],
    [1, 3, 3, 5],
    default=np.nan
).astype(float)

# Skor tanah contoh: kode 1, 2, 3
soil_score = np.select(
    [soil == 1, soil == 2, soil == 3],
    [1, 3, 5],
    default=np.nan
).astype(float)

# Elevasi bukan penyebab langsung longsor.
# Skor ini hanya contoh untuk menunjukkan pengolahan DEM.
elevation_score = np.select(
    [dem < 200, dem < 500, dem < 800, dem < 1200],
    [1, 2, 3, 4],
    default=5
).astype(float)

print("\n4. KLASIFIKASI PARAMETER")
print("Slope (derajat): <8, 8-15, 15-25, 25-40, >=40")
print("Hujan (mm/tahun): <1500, 1500-2000, 2000-2500,")
print("                  2500-3000, >=3000")
print("Skor akhir setiap parameter: 1 (rendah) hingga 5 (tinggi).")
print("Catatan: batas kelas di atas merupakan contoh pembelajaran.")

# ------------------------------------------------------------
# 6. WEIGHTED OVERLAY
# Formula: skor = jumlah (skor parameter x bobot parameter)
# ------------------------------------------------------------

score_stack = np.stack([
    slope_score,
    rain_score,
    land_score,
    soil_score,
    elevation_score
])

weight_array = np.array([
    weights["slope"],
    weights["rainfall"],
    weights["landuse"],
    weights["soil"],
    weights["elevation"]
]).reshape(-1, 1, 1)

valid = np.all(np.isfinite(score_stack), axis=0)

final_score = np.full(dem.shape, np.nan, dtype=float)
final_score[valid] = np.sum(
    score_stack * weight_array, axis=0
)[valid]

print("\n5. PEMBOBOTAN")
for name, weight in weights.items():
    print(f"{name:12s}: {weight * 100:.0f}%")

# ------------------------------------------------------------
# 7. ZONASI POTENSI KERAWANAN
# Karena skor input 1-5 dan bobot berjumlah 1,
# skor gabungan juga berada pada rentang 1-5.
# ------------------------------------------------------------

zone = np.full(dem.shape, np.nan, dtype=float)

zone[valid & (final_score < 2.0)] = 1
zone[valid & (final_score >= 2.0) &
     (final_score < 3.5)] = 2
zone[valid & (final_score >= 3.5)] = 3

zone_names = {
    1: "Rendah",
    2: "Sedang",
    3: "Tinggi"
}

print("\n6. HASIL ZONASI")
for code, name in zone_names.items():
    count = np.sum(zone == code)
    percentage = count / np.sum(valid) * 100 if np.sum(valid) else 0
    print(f"Potensi {name:7s}: {count} piksel ({percentage:.1f}%)")

print(
    "\nInterpretasi: zona tinggi memiliki kombinasi skor parameter "
    "yang relatif lebih mendukung kerawanan longsor."
)

# ------------------------------------------------------------
# 8. VISUALISASI PETA
# ------------------------------------------------------------

fig, axes = plt.subplots(2, 3, figsize=(15, 9))
plots = [
    (dem, "DEM (m)", "terrain"),
    (slope, "Slope (derajat)", "viridis"),
    (rainfall, "Curah hujan (mm/tahun)", "Blues"),
    (land_score, "Skor penggunaan lahan", "YlOrBr"),
    (soil_score, "Skor jenis tanah", "YlOrBr"),
    (final_score, "Skor weighted overlay", "YlOrRd")
]

for ax, (array, title, cmap) in zip(axes.flat, plots):
    im = ax.imshow(array, cmap=cmap)
    ax.set_title(title)
    ax.set_xticks([])
    ax.set_yticks([])
    plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

plt.suptitle("Parameter Analisis Potensi Kerawanan Longsor")
plt.tight_layout()
plt.show()

# Peta zonasi
from matplotlib.colors import ListedColormap
zone_cmap = ListedColormap(["#4daf4a", "#ffcc33", "#d7191c"])

plt.figure(figsize=(8, 6))
plt.imshow(zone, cmap=zone_cmap, vmin=1, vmax=3)
plt.title("Zonasi Potensi Kerawanan Longsor")
plt.xticks([])
plt.yticks([])
cbar = plt.colorbar(ticks=[1, 2, 3])
cbar.ax.set_yticklabels(["Rendah", "Sedang", "Tinggi"])
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 9. VALIDASI
# Jika tersedia titik kejadian longsor, validasi perlu dilakukan
# dengan koordinat kejadian yang sesuai dengan raster.
# Kode ini belum menghitung akurasi tanpa data kejadian.
# ------------------------------------------------------------

print("\n7. VALIDASI")
print(
    "Bandingkan peta zonasi dengan inventaris kejadian longsor "
    "aktual. Periksa apakah titik kejadian lebih banyak berada "
    "pada zona dengan skor tinggi. Gunakan data validasi terpisah "
    "bila menghitung metrik akurasi."
)

print("\n8. KETERBATASAN")
print(
    "Bobot dan ambang kelas adalah contoh. Hasil dipengaruhi "
    "kualitas raster, skala peta, kondisi geologi, dan data kejadian. "
    "Peta menunjukkan potensi relatif, bukan kepastian longsor."
)

print("\n9. KESIMPULAN")
print(
    "GIS menggabungkan DEM, slope, curah hujan, penggunaan lahan, "
    "dan jenis tanah melalui reclassification dan weighted overlay "
    "untuk menghasilkan zonasi potensi kerawanan longsor."
)

# ------------------------------------------------------------
# 10. SIMPAN OUTPUT
# ------------------------------------------------------------

pd.DataFrame({
    "Parameter": list(weights.keys()),
    "Bobot (%)": [v * 100 for v in weights.values()]
}).to_csv("bobot_parameter.csv", index=False)

pd.DataFrame(final_score).to_csv(
    "skor_weighted_overlay.csv", index=False
)

pd.DataFrame(zone).to_csv(
    "zonasi_potensi_longsor.csv", index=False
)

print("\nFile CSV berhasil dibuat di folder kerja notebook.")
```

```python

# ============================================================
# ANALISIS DATA METEOROLOGI DAN CURAH HUJAN
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. MEMUAT DATA
# ------------------------------------------------------------

# True: menggunakan data simulasi untuk menguji program
# False: membaca file CSV milik sendiri
USE_DEMO = True

if USE_DEMO:
    np.random.seed(42)

    tanggal = pd.date_range(
        start="2026-01-01",
        periods=30,
        freq="D"
    )

    n = len(tanggal)

    data = pd.DataFrame({
        "Tanggal": tanggal,
        "Curah_Hujan_mm": np.random.gamma(1.5, 5, n),
        "Temperatur_C": np.random.normal(28, 2, n),
        "Kelembapan_persen": np.random.normal(78, 8, n),
        "Tekanan_hPa": np.random.normal(1008, 3, n),
        "Angin_m_s": np.random.uniform(0.5, 6, n)
    })

    # Contoh hari tanpa hujan
    data.loc[2, "Curah_Hujan_mm"] = 0
    data.loc[10, "Curah_Hujan_mm"] = 0

else:
    # Ganti nama file sesuai file CSV kamu.
    data = pd.read_csv("data_cuaca.csv")
    data["Tanggal"] = pd.to_datetime(
        data["Tanggal"], errors="coerce"
    )

# Nama kolom yang digunakan:
# Tanggal, Curah_Hujan_mm, Temperatur_C,
# Kelembapan_persen, Tekanan_hPa, Angin_m_s

parameter = [
    "Curah_Hujan_mm",
    "Temperatur_C",
    "Kelembapan_persen",
    "Tekanan_hPa",
    "Angin_m_s"
]

# ------------------------------------------------------------
# 2. DASAR TEORI
# ------------------------------------------------------------

print("=" * 60)
print("ANALISIS DATA METEOROLOGI")
print("=" * 60)

print("\n1. DASAR TEORI")
print("- Curah hujan: jumlah hujan yang terukur dalam mm.")
print("- Temperatur: tingkat panas atau dinginnya udara (°C).")
print("- Kelembapan relatif: persentase kandungan uap air (%).")
print("- Tekanan udara: tekanan atmosfer (hPa).")
print("- Kecepatan angin: laju gerakan udara (m/s).")

# ------------------------------------------------------------
# 3. DESKRIPSI DAN KUALITAS DATA
# ------------------------------------------------------------

print("\n2. DESKRIPSI DATA")
print("Jumlah baris:", len(data))
print("Periode:", data["Tanggal"].min(), "sampai",
      data["Tanggal"].max())
print("\nContoh data:")
print(data.head())

print("\n3. PEMERIKSAAN KUALITAS DATA")

# Mengubah kolom parameter menjadi numerik.
for kolom in parameter:
    data[kolom] = pd.to_numeric(
        data[kolom], errors="coerce"
    )

print("\nJumlah data kosong:")
print(data[["Tanggal"] + parameter].isna().sum())

print("\nJumlah tanggal tidak valid:",
      data["Tanggal"].isna().sum())

# Pemeriksaan nilai yang tidak masuk akal secara fisik.
# Ambang berikut merupakan pemeriksaan umum, bukan batas
# universal untuk semua alat dan kondisi stasiun.
batas_fisik = {
    "Curah_Hujan_mm": (0, 1000),
    "Temperatur_C": (-90, 65),
    "Kelembapan_persen": (0, 100),
    "Tekanan_hPa": (800, 1100),
    "Angin_m_s": (0, 150)
}

for kolom, (bawah, atas) in batas_fisik.items():
    invalid = (
        (data[kolom] < bawah) |
        (data[kolom] > atas)
    )
    print(f"{kolom}: {invalid.sum()} nilai di luar batas pemeriksaan")

# ------------------------------------------------------------
# 4. IDENTIFIKASI OUTLIER METODE IQR
# Data outlier ditandai, tidak langsung dihapus.
# ------------------------------------------------------------

print("\n4. PEMERIKSAAN OUTLIER (METODE IQR)")

outlier_summary = {}

for kolom in parameter:
    seri = data[kolom].dropna()

    if len(seri) == 0:
        outlier_summary[kolom] = 0
        continue

    q1 = seri.quantile(0.25)
    q3 = seri.quantile(0.75)
    iqr = q3 - q1

    batas_bawah = q1 - 1.5 * iqr
    batas_atas = q3 + 1.5 * iqr

    mask = (
        (data[kolom] < batas_bawah) |
        (data[kolom] > batas_atas)
    )

    outlier_summary[kolom] = int(mask.sum())

    print(
        f"{kolom}: {mask.sum()} outlier "
        f"(batas {batas_bawah:.2f} sampai {batas_atas:.2f})"
    )

print("Outlier tidak dihapus otomatis karena dapat berupa")
print("kejadian cuaca ekstrem yang benar-benar terjadi.")

# ------------------------------------------------------------
# 5. STATISTIK DASAR
# ------------------------------------------------------------

print("\n5. STATISTIK DASAR")

statistik = data[parameter].agg(
    ["mean", "min", "max", "median", "std"]
).T

statistik.columns = [
    "Rata-rata", "Minimum", "Maksimum", "Median", "Simpangan Baku"
]

print(statistik.round(2))

# ------------------------------------------------------------
# 6. ANALISIS CURAH HUJAN
# ------------------------------------------------------------

print("\n6. ANALISIS CURAH HUJAN")

hujan_valid = data["Curah_Hujan_mm"].dropna()

print("Total curah hujan (mm):",
      round(hujan_valid.sum(), 2))

print("Rata-rata curah hujan per pengamatan (mm):",
      round(hujan_valid.mean(), 2))

if len(hujan_valid) > 0:
    idx_max = data["Curah_Hujan_mm"].idxmax()

    print("Curah hujan tertinggi (mm):",
          data.loc[idx_max, "Curah_Hujan_mm"])

    print("Waktu curah hujan tertinggi:",
          data.loc[idx_max, "Tanggal"])

    print("Jumlah hari/pengamatan tanpa hujan:",
          (data["Curah_Hujan_mm"] == 0).sum())

# ------------------------------------------------------------
# 7. HUBUNGAN ANTARPARAMETER
# ------------------------------------------------------------

print("\n7. KORELASI ANTARPARAMETER")

kolom_korelasi = [
    "Curah_Hujan_mm",
    "Temperatur_C",
    "Kelembapan_persen",
    "Tekanan_hPa",
    "Angin_m_s"
]

korelasi = data[kolom_korelasi].corr(method="pearson")
print(korelasi.round(2))

r_hujan_kelembapan = data[
    ["Curah_Hujan_mm", "Kelembapan_persen"]
].corr().iloc[0, 1]

r_hujan_temperatur = data[
    ["Curah_Hujan_mm", "Temperatur_C"]
].corr().iloc[0, 1]

print("\nKorelasi hujan dengan kelembapan:",
      round(r_hujan_kelembapan, 3))

print("Korelasi hujan dengan temperatur:",
      round(r_hujan_temperatur, 3))

def interpretasi_korelasi(r):
    if pd.isna(r):
        return "Tidak dapat dihitung."
    nilai = abs(r)

    if nilai < 0.2:
        kekuatan = "sangat lemah"
    elif nilai < 0.4:
        kekuatan = "lemah"
    elif nilai < 0.6:
        kekuatan = "sedang"
    elif nilai < 0.8:
        kekuatan = "kuat"
    else:
        kekuatan = "sangat kuat"

    arah = "positif" if r > 0 else "negatif" if r < 0 else "nol"
    return f"Korelasi {arah}, dengan kekuatan {kekuatan}."

print(interpretasi_korelasi(r_hujan_kelembapan))
print(interpretasi_korelasi(r_hujan_temperatur))

print(
    "Korelasi menunjukkan keterkaitan statistik, bukan bukti "
    "bahwa satu parameter menyebabkan perubahan parameter lain."
)

# ------------------------------------------------------------
# 8. VISUALISASI PERUBAHAN TERHADAP WAKTU
# ------------------------------------------------------------

data_plot = data.sort_values("Tanggal")

fig, axes = plt.subplots(3, 1, figsize=(12, 10), sharex=True)

axes[0].bar(
    data_plot["Tanggal"],
    data_plot["Curah_Hujan_mm"]
)
axes[0].set_ylabel("Hujan (mm)")
axes[0].set_title("Curah Hujan Harian")
axes[0].grid(True, alpha=0.3)

axes[1].plot(
    data_plot["Tanggal"],
    data_plot["Temperatur_C"],
    label="Temperatur (°C)"
)
axes[1].plot(
    data_plot["Tanggal"],
    data_plot["Kelembapan_persen"],
    label="Kelembapan (%)"
)
axes[1].set_ylabel("Nilai parameter")
axes[1].legend()
axes[1].grid(True, alpha=0.3)

axes[2].plot(
    data_plot["Tanggal"],
    data_plot["Tekanan_hPa"],
    label="Tekanan (hPa)"
)
axes[2].plot(
    data_plot["Tanggal"],
    data_plot["Angin_m_s"],
    label="Angin (m/s)"
)
axes[2].set_ylabel("Nilai parameter")
axes[2].set_xlabel("Tanggal")
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 9. SCATTER PLOT HUBUNGAN HUJAN
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].scatter(
    data["Kelembapan_persen"],
    data["Curah_Hujan_mm"],
    alpha=0.7
)
axes[0].set_xlabel("Kelembapan relatif (%)")
axes[0].set_ylabel("Curah hujan (mm)")
axes[0].set_title("Hujan vs Kelembapan")
axes[0].grid(True, alpha=0.3)

axes[1].scatter(
    data["Temperatur_C"],
    data["Curah_Hujan_mm"],
    alpha=0.7
)
axes[1].set_xlabel("Temperatur (°C)")
axes[1].set_ylabel("Curah hujan (mm)")
axes[1].set_title("Hujan vs Temperatur")
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 10. INTERPRETASI OTOMATIS
# ------------------------------------------------------------

print("\n8. INTERPRETASI METEOROLOGI")

if len(hujan_valid) > 0:
    print(
        f"Total hujan selama periode pengamatan adalah "
        f"{hujan_valid.sum():.2f} mm."
    )

    print(
        f"Rata-rata hujan per pengamatan adalah "
        f"{hujan_valid.mean():.2f} mm."
    )

print(
    "Periksa grafik untuk melihat perubahan hujan, temperatur, "
    "kelembapan, tekanan, dan angin sepanjang waktu."
)

print(
    "Korelasi positif berarti dua variabel cenderung meningkat "
    "bersama, sedangkan korelasi negatif berarti satu variabel "
    "cenderung menurun saat variabel lainnya meningkat."
)

print("\n9. KETERBATASAN")
print(
    "Kesimpulan dibatasi oleh panjang periode, kelengkapan data, "
    "ketelitian alat, dan karakteristik stasiun. Outlier statistik "
    "perlu diperiksa sebelum diputuskan sebagai kesalahan data."
)

print("\n10. KESIMPULAN")
print(
    "Analisis statistik dan grafik digunakan untuk memahami pola "
    "cuaca. Korelasi tidak membuktikan hubungan sebab-akibat."
)

# ------------------------------------------------------------
# 11. SIMPAN HASIL
# ------------------------------------------------------------

data.to_csv("data_meteorologi_bersih.csv", index=False)
statistik.to_csv("statistik_meteorologi.csv")
korelasi.to_csv("korelasi_meteorologi.csv")

print("\nFile CSV berhasil dibuat:")
print("- data_meteorologi_bersih.csv")
print("- statistik_meteorologi.csv")
print("- korelasi_meteorologi.csv")
```

Analisis meteorologi dilakukan untuk mengetahui kondisi cuaca berdasarkan curah hujan, temperatur, kelembapan, tekanan udara, dan kecepatan angin. Pemeriksaan data dilakukan untuk menemukan nilai kosong dan pencilan (outlier). ```python

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress

# 1. DATA SIMULASI
# Waktu dihitung dalam detik dari acuan waktu yang sama
data = {
    "Stasiun": ["STA01", "STA02", "STA03", "STA04",
                "STA05", "STA06", "STA07", "STA08"],
    "Tp_s": [12, 18, 24, 30, 36, 42, 48, 54]
}

df = pd.DataFrame(data)

# Parameter simulasi, bukan hasil pengamatan nyata
T0_sebenarnya = 5.0
Vp_vs_sebenarnya = 1.75

# Membuat waktu S-P dengan sedikit variasi simulasi
noise = np.array([0.03, -0.04, 0.02, -0.02,
                  0.01, 0.04, -0.03, 0.00])

df["S_minus_P_s"] = (
    (df["Tp_s"] - T0_sebenarnya)
    * (Vp_vs_sebenarnya - 1)
    / Vp_vs_sebenarnya
    + noise
)

df["Ts_s"] = df["Tp_s"] + df["S_minus_P_s"]

# 2. TABEL DATA
print("TABEL WAKTU TIBA DAN SELISIH S-P")
print(df.round(3).to_string(index=False))

# 3. REGRESI WADATI
x = df["Tp_s"].to_numpy()
y = df["S_minus_P_s"].to_numpy()

hasil = linregress(x, y)

m = hasil.slope
b = hasil.intercept
r2 = hasil.rvalue ** 2

vp_vs = m + 1
T0 = -b / m

print("\nHASIL ANALISIS")
print(f"Persamaan regresi: Y = {m:.4f}X + ({b:.4f})")
print(f"Kemiringan garis: {m:.4f}")
print(f"Intersep: {b:.4f} detik")
print(f"R²: {r2:.4f}")
print(f"Rasio Vp/Vs: {vp_vs:.4f}")
print(f"Perkiraan waktu asal: {T0:.3f} detik dari acuan waktu")

# 4. GRAFIK DIAGRAM WADATI
x_garis = np.linspace(x.min(), x.max(), 200)
y_garis = m * x_garis + b

plt.figure(figsize=(9, 6))
plt.scatter(x, y, label="Data simulasi")
plt.plot(x_garis, y_garis, label="Regresi linear")

for _, baris in df.iterrows():
    plt.annotate(
        baris["Stasiun"],
        (baris["Tp_s"], baris["S_minus_P_s"]),
        xytext=(4, 5),
        textcoords="offset points",
        fontsize=8
    )

plt.xlabel("Waktu tiba gelombang P, Tp (detik)")
plt.ylabel("Selisih waktu S-P (detik)")
plt.title("Diagram Wadati - Data Simulasi")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# 5. INTERPRETASI OTOMATIS
print("\nINTERPRETASI")
print(
    "Diagram memperlihatkan hubungan antara waktu tiba P "
    "dan selisih waktu S-P."
)
print(
    f"Kemiringan regresi menghasilkan rasio Vp/Vs sekitar "
    f"{vp_vs:.3f}."
)
print(
    f"Waktu asal gempa diperkirakan {T0:.3f} detik setelah "
    "acuan waktu simulasi."
)
print(
    "Hasil ini berasal dari data simulasi untuk pembelajaran, "
    "bukan parameter gempa nyata."
)
```

Berdasarkan analisis data simulasi, Diagram Wadati menunjukkan hubungan linear antara waktu tiba gelombang P dan selisih waktu tiba gelombang S–P. Semakin lama waktu tiba gelombang P, selisih waktu S–P cenderung semakin besar.
Garis regresi digunakan untuk memperkirakan waktu asal gempa (origin time) dan rasio kecepatan gelombang \(Vp/Vs\). Waktu asal diperoleh dari perpotongan garis regresi dengan sumbu waktu, sedangkan rasio kecepatan dihitung dari kemiringan garis ditambah satu.
Hasil analisis ini menggambarkan cara kerja Diagram Wadati dalam menentukan parameter dasar gempa. Namun, karena data yang digunakan merupakan data simulasi, hasilnya belum mewakili kejadian gempa nyata. Pada data sebenarnya, ketelitian hasil dipengaruhi oleh pembacaan waktu tiba gelombang P dan S serta kondisi medium yang dilalui gelombang.

```python

# ==========================================================
# PEMETAAN CUACA GLOBAL MENGGUNAKAN PYTHON
# Data simulasi, bukan pengamatan cuaca aktual
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt

# 1. MEMBUAT KOORDINAT GLOBAL
latitude = np.arange(-90, 91, 2)
longitude = np.arange(-180, 181, 2)

Lon, Lat = np.meshgrid(longitude, latitude)

# 2. MEMBUAT DATA TEMPERATUR SIMULASI
# Ilustrasi: temperatur cenderung lebih tinggi di ekuator
# dan lebih rendah mendekati kutub.

np.random.seed(10)

temperatur = (
    30
    - 0.55 * np.abs(Lat)
    + 3 * np.cos(np.radians(Lon)) * np.cos(np.radians(Lat))
    + 1.5 * np.random.randn(*Lat.shape)
)

# 3. VALIDASI DATA
assert np.all((-90 <= Lat) & (Lat <= 90))
assert np.all((-180 <= Lon) & (Lon <= 180))
assert Lon.shape == Lat.shape == temperatur.shape
assert np.all(np.isfinite(temperatur))

# 4. MEMBUAT PETA TEMPERATUR GLOBAL
plt.figure(figsize=(14, 6))

peta = plt.pcolormesh(
    Lon,
    Lat,
    temperatur,
    shading="auto",
    cmap="turbo"
)

colorbar = plt.colorbar(peta)
colorbar.set_label("Temperatur (°C)", fontsize=11)

plt.xlabel("Longitude (derajat)")
plt.ylabel("Latitude (derajat)")
plt.title(
    "Peta Temperatur Global\n"
    "Data Simulasi - Bukan Pengamatan Aktual"
)

plt.xticks(np.arange(-180, 181, 60))
plt.yticks(np.arange(-90, 91, 30))
plt.xlim(-180, 180)
plt.ylim(-90, 90)
plt.grid(alpha=0.25)
plt.tight_layout()

# Simpan gambar
plt.savefig(
    "peta_temperatur_global.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# 5. STATISTIK TEMPERATUR
print("=== STATISTIK TEMPERATUR SIMULASI ===")
print(f"Temperatur minimum : {temperatur.min():.2f} °C")
print(f"Temperatur maksimum: {temperatur.max():.2f} °C")
print(f"Temperatur rata-rata: {temperatur.mean():.2f} °C")

# 6. INTERPRETASI OTOMATIS
print("\n=== INTERPRETASI ===")
print(
    "Peta menunjukkan variasi temperatur simulasi "
    "berdasarkan koordinat global."
)
print(
    "Pola ilustrasi memperlihatkan temperatur yang cenderung "
    "lebih tinggi di sekitar ekuator dan lebih rendah "
    "mendekati kutub."
)
print(
    "Data ini hanya simulasi, sehingga tidak menggambarkan "
    "temperatur dunia yang sebenarnya."
)
print("\nGambar tersimpan: peta_temperatur_global.png")
```

Berdasarkan hasil visualisasi menggunakan Python, peta menunjukkan persebaran temperatur simulasi berdasarkan koordinat latitude dan longitude. Warna pada peta merepresentasikan variasi temperatur, sedangkan colorbar digunakan untuk mengetahui nilai temperatur dalam derajat Celsius (°C).
Pola simulasi memperlihatkan temperatur yang cenderung lebih tinggi di sekitar ekuator dan lebih rendah mendekati kutub. Pola ini menggambarkan pengaruh perbedaan penerimaan energi Matahari berdasarkan lintang geografis.
